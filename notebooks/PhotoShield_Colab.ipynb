```python
# 3️⃣ Change to the project directory
%cd PhotoSheildAntigravity
```

```python
# 4️⃣ Start the FastAPI server in the background (listens on 0.0.0.0:8000)
import subprocess, time, os, signal

server = subprocess.Popen(
    ["uvicorn", "backend.main:app", "--host", "0.0.0.0", "--port", "8000"],
    cwd=os.getcwd()
)

# give the server a couple of seconds to spin up
time.sleep(5)
print("🚀 FastAPI server started")
```

```python
# 5️⃣ Open a public tunnel (ngrok alternative – pyngrok)
from pyngrok import ngrok

public_url = ngrok.connect(8000, "http")
print("🔗 Backend reachable at:", public_url)

# Store the URL for later Streamlit launch
os.environ["BACKEND_URL"] = str(public_url)
```

```python
# 6️⃣ Launch the Streamlit UI (also behind the same tunnel)
import os, subprocess, time

# Streamlit will read BACKEND_URL from the env
ui = subprocess.Popen(["streamlit", "run", "frontend/app.py", "--server.port", "8501"], env=os.environ.copy())
time.sleep(5)

ui_url = ngrok.connect(8501, "http")
print("💻 Streamlit UI reachable at:", ui_url)

print("✅ All set – open the UI link above, upload a picture, and watch the shield in action!")
```

### 📌 Quick demo script (2 min)

1. Click the **Streamlit UI** link printed in the last cell.
2. Upload any portrait (≤ 1 MB).
3. Adjust the **Strength** slider (default 0.02) – higher values give stronger protection but may lower visual fidelity.
4. Press **Generate Shield** (≈ 5 s on a free T4).
5. Click **Run Inpaint demo** with a simple prompt like *"a person wearing a blue hat"*.
6. Observe the **Original edit** vs. **Shielded edit** side‑by‑side.
7. Switch to the **Detector** tab to see the ELA heat‑map and the fake‑confidence score.
8. Hit **Download PDF** on the **Report** section to generate a ready‑to‑submit evidence bundle.

All steps fit comfortably under a minute per image on the Colab T4.
